```python
import os
import sys

# 1. Dynamically locate the 'src' directory relative to this notebook
notebook_dir = os.path.dirname(os.path.abspath("__file__")) if "__file__" in globals() else os.getcwd()
src_dir = os.path.abspath(os.path.join(notebook_dir, "..", "src"))

# 2. Add 'src' to Python's search path if not already present
if src_dir not in sys.path:
    sys.path.append(src_dir)

# 3. Now import your domain class
from project1_population import PopulationModel

import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
rng = np.random.default_rng(seed=42)
print("Setup complete!")
```

```python
base_year = 2024
initial_pop = 45.5  # Uganda population in millions (illustrative/UBOS)
years_to_project = 26

model = PopulationModel(base_year=base_year, base_population=initial_pop)
exp_pop = model.exponential_growth(rate=0.032, years=years_to_project)
log_pop = model.logistic_growth(rate=0.032, capacity=120.0, years=years_to_project)

time_axis = np.arange(base_year, base_year + years_to_project + 1)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(time_axis, exp_pop, label="Exponential Growth (r=3.2%)", linestyle="--", color="crimson")
plt.plot(time_axis, log_pop, label="Logistic Growth (K=120M)", linewidth=2, color="navy")
plt.title("Uganda Population Projection (2024 - 2050)")
plt.xlabel("Year")
plt.ylabel("Population (Millions)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
